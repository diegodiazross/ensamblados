## 4 Ensamblados secuenciales de impulso de gradiente.

Para terminar nuestra breve taxonomía de los métodos de ensamblado hablaremos  de los ensamblados secuenciales de impulso de gradiente y daremos un ejemplo práctico de impulso de gradiente newtoniano que es considerado el estado del arte dentro de los métodos de ensamblado.

Ya mencionamos anteriormente que en los ensamblados secuenciales cada uno de los modelos se concentra en los errores cometidos por el modelo anterior. La diferencia entre los secuenciales de impulso adaptativo y los de impulso de gradiente es que estos últimos utilizan la derivada de la función de pérdida con respecto a nuestro predictor  (descenso de gradiente y/o método de newton ) mientras que en el impulso adaptativo se utilizan pesos de acuerdo al error cometido por el modelo base.


A continuación daremos un ejemplo práctico de impulso de gradiente newtoniano utilizando los datos de clasificación de correo en spam que también utilizamos en el ejemplo de impulso adaptativo.  Debemos  instalar xgboost, que es el framework que implementa impulso newtoniano y otras mejoras algorítmicas.  


```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split,GridSearchCV
from sklearn.metrics import accuracy_score,f1_score ,classification_report
from xgboost import XGBClassifier
```

```python
df = pd.read_csv('data/spambase.data', header=None)
```

```python
X = df.iloc[:, :-1]
Y = df.iloc[:, -1]
```

```python
# estandarizacion de los datos
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
params = {'max': [1, 2, 3, 4, 5],
          'learning_rate': np.linspace(0, 1, num=10),
          'reg_alpha': [0, 0.1, 0.5, 1.0, 3, 5, 10],
          'reg_lambda': [0, 0.1, 0.5, 1.0, 3, 5, 10],
         }
```

```python
ens = XGBClassifier(booster='gbtree', objective='binary:logistic', eval_metric='logloss',n_jobs=-1)
```

```python
clf = GridSearchCV(estimator=ens, param_grid=params, cv=5, n_jobs=-1)
```

```python
clf.fit(X_scaled, Y)
```
